```python
import pandas as pd

logon_path = "../data/raw/r1/r1/logon.csv"

logon = pd.read_csv(logon_path)

print("Logon shape:", logon.shape)
print("\nColumns:")
print(logon.columns.tolist())

print("\nFirst 5 rows:")
display(logon.head())
```

```python
print("Activity values:")
print(logon["activity"].value_counts())

print("\nNumber of users:")
print(logon["user"].nunique())

print("\nNumber of PCs:")
print(logon["pc"].nunique())

print("\nDate range:")
print(logon["date"].min(), "to", logon["date"].max())
```

```python
logon["date"] = pd.to_datetime(
    logon["date"],
    format="%m/%d/%Y %H:%M:%S"
)

print("Correct date range:")
print(logon["date"].min())
print("to")
print(logon["date"].max())
```

```python
print("Missing values:")
print(logon.isnull().sum())

print("\nDuplicate rows:")
print(logon.duplicated().sum())
```

```python
print("\nData types:")
print(logon.dtypes)
```
